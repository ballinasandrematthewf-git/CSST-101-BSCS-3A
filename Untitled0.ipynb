Apply the rule to the multiple student

```r

student_risk <- function(attendance, grade) {
  if (attendance < 75 & grade < 75) {
    return("High Risk")
  } else {
    return("Low Risk")
  }
}

students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)


students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

students
```

How many students are High Risk?

3 (Students S001, S003, and S004)

Which group has the largest number of students?

What does the graph show?
It shows the frequency (number of students) falling under each risk category ("High Risk" vs. "Low Risk").

Which risk category has the most students?
Low Risk (with 3 students, compared to 2 in High Risk).

What action should the university take?
Provide targeted academic intervention and attendance support to the High Risk students (S002 and S005) to help them improve their performance.

Add recommendation

```r
get_recommendation <- function(risk_level) {
  if (risk_level == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk_level == "Moderate Risk") {
    return("Academic monitoring")
  } else {
    return("Continue regular monitoring")
  }
}


students$Recommendation <- sapply(students$Risk, get_recommendation)


students
```

```r

generate_report <- function(id, attendance, grade) {


  risk_level <- ifelse(attendance < 75 & grade < 75, "High Risk",
                ifelse(attendance < 75 | grade < 75, "Moderate Risk", "Low Risk"))

  recommendation <- ifelse(risk_level == "High Risk", "Immediate academic intervention",
                    ifelse(risk_level == "Moderate Risk", "Academic monitoring",
                           "Continue regular monitoring"))


  cat("Student ID:", id, "\n")
  cat("Attendance:", paste0(attendance, "%"), "\n")
  cat("Grade:", grade, "\n\n")
  cat("Risk Level:", risk_level, "\n\n")
  cat("Recommendation:\n")
  cat(recommendation, "\n")
}


generate_report("S002", 72, 70)
```
